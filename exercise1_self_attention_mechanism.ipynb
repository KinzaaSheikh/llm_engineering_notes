```python
import numpy as np

# Toy sentence
tokens = ["The", "cat", "sat"]

# Toy embeddings
# Each token is represented by a 4-dimensional vector.
embeddings = np.array([
    [ 0.8, -0.2,  0.4,  0.1],   # The
    [ 0.7,  0.1,  0.5, -0.3],   # cat
    [-0.1,  0.2,  0.0,  0.9]    # sat
])

print("Tokens:", tokens)
print("Embedding matrix:\n", embeddings)
print("Shape:", embeddings.shape)
```

```python
# Simplification for this learning exercise:
# Q and K are currently the same as our embeddings.

Q = embeddings
K = embeddings

print("Q shape:", Q.shape)
print("K shape:", K.shape)
```

```python
# Compare "The" (query) with "cat" (key)

query = Q[0]
key = K[1]

score = query @ key

print("Query:", query)
print("Key:", key)
print("Dot-product score:", score)
```

```python
# Q @ K.T calculates every query-key dot product at once.

scores = Q @ K.T

print("Raw attention scores:")
print(scores)
```

```python
# The dimensionality of each key vector
d_k = K.shape[1]

# Scale by sqrt(d_k)
scaled_scores = scores / np.sqrt(d_k)

print("d_k:", d_k)
print("sqrt(d_k):", np.sqrt(d_k))
print("\nScaled attention scores:")
print(scaled_scores)
```

```python
print("Embeddings:", embeddings.shape)
print("Q:", Q.shape)
print("K:", K.shape)
print("K.T:", K.T.shape)
print("Q @ K.T:", (Q @ K.T).shape)
print("Scaled scores:", scaled_scores.shape)
```

```python
def softmax(x):
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)


attention_weights = softmax(scaled_scores)

print("Attention weights:")
print(attention_weights)

print("\nRow sums:")
print(attention_weights.sum(axis=-1))
```

Compare every query with every key → stabilize the scores → turn them into attention weights.


```python
V = embeddings

print("V:")
print(V)
print("V shape:", V.shape)
```

```python
attention_output = attention_weights @ V

print("Attention output:")
print(attention_output)

print("Shape:", attention_output.shape)
```
